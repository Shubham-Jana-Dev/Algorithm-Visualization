```python
class Greads:
  def student_gread():
    gread = int(input("Enter your greads 📑: "))
    if(100>=gread>=90):
      print("Nice you got A+ 🎉")
    elif(90>gread>=80):
      print("Good you got A ✨")
    elif(80>gread>=70):
      print("Not Bad you got B 🎊")
    elif(70>gread>=60):
      print("Ok you got C 🎀")
    elif(60>gread>=50):
      print("Oh god you got D 🤔")
    elif(50>gread>=0):
      print("Focus on study not ... 🤬")
    else:
      print("Invalid input :(")
g = Greads
g.student_gread()
```

```python
class CalTax:
  def calculate_your_tax():
    income = float(input("Enter your salary: "))
    if(income>500000):
      print(f"You have to pay {income * (25/100)} as your in come tax")
    elif(income>100000):
      print(f"You have to pay {income * (15/100)} as your in come tax")
    elif(income>90000):
      print(f"You have to pay {income * (10/100)} as your in come tax")
    elif(income>50000):
      print(f"You have to pay {income * (8/100)} as your in come tax")
    elif(0<income>40000):
      print(f"You have to pay {income * (5/100)} as your in come tax")
    else:
      print("Invalid input :(")
ct = CalTax
ct.calculate_your_tax()
```

```python
class ATM:
  def __init__(self,balance):
    self.balance = balance
  def withdrawal(self,amount):
    if(0<amount<=self.balance):
      print("collect your chase your remaining balance is:",self.balance-amount)
    else:
      print("Insaficient Balance :(")
  def diposit(self,amount):
    if(0<amount):
      print("current balance is:",self.balance+amount)
    else:
      print("Invalid amount")
Atm = ATM(4000)
Atm.withdrawal(200)
Atm.diposit(200)
```
