```python
import pandas as pd
```

```python
data = {
    "Fruits":["Apple","Banana","Orange","Grapes","Papaya"],
    "Cost":[200,300,240,100,80]
}
```

```python
df = pd.DataFrame(data)
```

```python
display(df)
```

```python
df.rename(columns={"Cost":"Price"},inplace=True)
```
