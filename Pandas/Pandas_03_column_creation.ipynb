```python
import pandas as pd
```

```python
my_data = {
    "Name":["Shubham","Deepak","Sujoy","Sujoy","Danosh","Pradip","Raj",None],
    "Salary":[34000,45000,25000,25000,20000,55000,None,90454]
}
```

```python
df = pd.DataFrame(my_data)
```

```python
display(df)
```

```python
df["Bonus"] = df["Salary"]*2
```

```python
display(df)
```

```python
df["Emplyee Id"] = pd.Series([34,67,23,23,66,12])
```

```python
display(df)
```

```python
df = df.sort_values(by='Emplyee Id',ascending=True)
```

```python
display(df)
```

```python
df = df.drop_duplicates(subset="Name",keep='first')
```

```python
display(df)
```

```python
df = df.dropna(subset="Name")
```

```python
display(df)
```
